### 1. Chứng Minh Toán Học Của Phương Pháp Gradient Descent

PHẦN 1: Chứng minh hướng "Trừ Độ Dốc" ($- \text{Độ dốc}$) là hướng đi xuống nhanh nhất
<br/>1.1. Đặt bài toán
Cho tập dữ liệu huấn luyện gồm $n$ mẫu: $\mathcal{D} = {(x_i, y_i)}_{i=1}^n$, với $x_i$ là biến đầu vào và $y_i$ là biến đầu ra thực tế.

Sai số tại từng điểm dữ liệu: $$e_i = y_i - f(x_i, \theta)$$
Hàm mất mát trung bình bình phương sai số (Mean Squared Error - MSE): $$E(\theta) = \frac{1}{n} \sum_{i=1}^n \left[ y_i - f(x_i, \theta) \right]^2$$
Mục tiêu tối ưu: $$\hat{\theta} = \arg\min_\theta E(\theta)$$

1.2. Chứng minh hàm mục tiêu giảm dần qua mỗi bước lặp Gradient Descent
Xét bài toán cực tiểu hóa hàm khả vi $f(x) \to \min$. Thuật toán Gradient Descent cập nhật tham số theo công thức: $$x_{k+1} = x_k - \eta \nabla f(x_k) \quad (\text{với } \eta > 0 \text{ là tốc độ học})$$

Bước 1: Khai triển Taylor bậc 1
Xấp xỉ giá trị của $f(x)$ tại điểm $x_{k+1}$ lân cận điểm $x_k$: $$f(x_{k+1}) \approx f(x_k) + \nabla f(x_k)^T (x_{k+1} - x_k)$$

Bước 2: Thay quy tắc cập nhật vào khai triển
Từ $x_{k+1} - x_k = -\eta \nabla f(x_k)$, ta có: $$f(x_{k+1}) \approx f(x_k) + \nabla f(x_k)^T \left( -\eta \nabla f(x_k) \right)$$ $$f(x_{k+1}) \approx f(x_k) - \eta |\nabla f(x_k)|^2$$

Bước 3: Đánh giá bất đẳng thức
Vì:

Tốc độ học $\eta > 0$
Bình phương chuẩn Euclid $|\nabla f(x_k)|^2 \ge 0$
Nên lượng trừ $-\eta |\nabla f(x_k)|^2 \le 0$. Do đó: $$f(x_{k+1}) \le f(x_k)$$

Kết luận: Giá trị của hàm mục tiêu luôn giảm dần hoặc giữ nguyên qua từng bước lặp ($f(x_0) \ge f(x_1) \ge f(x_2) \ge \dots$), đảm bảo dãy ${x_k}$ tiến dần về điểm cực tiểu cục bộ.

2. Chứng minh nghiệm chính xác (Phương pháp bình phương tối thiểu - OLS)
Xét mô hình hồi quy tuyến tính đơn biến: $$\hat{y} = f(x) = b + w x$$

Hàm tổn thất cần cực tiểu hóa: $$E(b, w) = \frac{1}{n} \sum_{i=1}^n \left[ y_i - (b + w x_i) \right]^2 \to \min$$

2.1. Đạo hàm riêng bậc nhất
Áp dụng quy tắc đạo hàm hàm hợp:

Đạo hàm theo $b$: $$\frac{\partial E}{\partial b} = \frac{1}{n} \sum_{i=1}^n 2\left[ y_i - (b + w x_i) \right] \cdot (-1) = -\frac{2}{n} \sum_{i=1}^n \left( y_i - b - w x_i \right)$$

Đạo hàm theo $w$: $$\frac{\partial E}{\partial w} = \frac{1}{n} \sum_{i=1}^n 2\left[ y_i - (b + w x_i) \right] \cdot (-x_i) = -\frac{2}{n} \sum_{i=1}^n x_i \left( y_i - b - w x_i \right)$$

Cho các đạo hàm riêng bằng $0$ để tìm điểm cực tiểu:

$$\begin{cases} \sum_{i=1}^n (y_i - b - w x_i) = 0 \ \sum_{i=1}^n x_i (y_i - b - w x_i) = 0 \end{cases} \implies \begin{cases} n b + w \sum_{i=1}^n x_i = \sum_{i=1}^n y_i & (1) \ b \sum_{i=1}^n x_i + w \sum_{i=1}^n x_i^2 = \sum_{i=1}^n x_i y_i & (2) \end{cases}$$

Giải hệ phương trình tìm $w$ và $b$:
Từ phương trình $(1)$, chia hai vế cho $n$ ta được: $$b = \bar{y} - w \bar{x}$$ (với $\bar{x}, \bar{y}$ là trung bình cộng của $x$ và $y$).

Thế $b = \bar{y} - w \bar{x}$ vào phương trình $(2)$ và rút gọn, ta thu được: $$w = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2}$$

Kết quả cuối cùng:
$$\begin{cases} w = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2} \ b = \bar{y} - w \bar{x} \end{cases}$$

<h1 style= "font-size : 30px; ">2. Viết chương trình tính $w, b$ (Nghiệm chính xác - Phương pháp bình phương tối thiểu) </h1>

Bước 1: Tính trung bình cộng của $x$ và $y$
$\bar{x} = \frac{1 + 2 + 3}{3} = 2$
$\bar{y} = \frac{2 + 4 + 5}{3} = 3.67 \quad (\text{hoặc } \frac{11}{3})$
Bước 2: Tìm mảng sai lệch (tương đương x1 = x - x_bar và y1 = y - y_bar trong code)
$x - \bar{x} = [1 - 2, ; 2 - 2, ; 3 - 2] = [-1, ; 0, ; 1]$
$y - \bar{y} = \left[2 - \frac{11}{3}, ; 4 - \frac{11}{3}, ; 5 - \frac{11}{3}\right] = \left[-\frac{5}{3}, ; \frac{1}{3}, ; \frac{4}{3}\right]$
Bước 3: Tính Tử số và Mẫu số của $w$
Tử số (Tổng tích chéo): $$\sum (x_i - \bar{x})(y_i - \bar{y}) = (-1) \cdot \left(-\frac{5}{3}\right) + 0 \cdot \frac{1}{3} + 1 \cdot \frac{4}{3} = \frac{5}{3} + \frac{4}{3} = 3$$

Mẫu số (Tổng bình phương sai lệch của $x$): $$\sum (x_i - \bar{x})^2 = (-1)^2 + 0^2 + 1^2 = 2$$

Bước 4: Tính hệ số $w$ và $b$
Hệ số góc $w$: $$w = \frac{\text{Tử số}}{\text{Mẫu số}} = \frac{3}{2} = 1.5$$
Hệ số chặn $b$: $$b = \bar{y} - w\bar{x} = \frac{11}{3} - 1.5 \cdot 2 = \frac{11}{3} - 3 = \frac{2}{3} \approx 0.67$$
Mô hình thu được: $y = 1.5x + 0.67$

```python
import numpy as np
import pandas as pd
```

```python
# Công thức tính 
def min_least_squares(x, y):
    w = np.sum((x - x.mean()) * (y - y.mean())) / np.sum((x - x.mean())**2)
    b = y.mean() - w * x.mean()
    return b, w
```

```python
x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

b, w = min_least_squares(x, y)
print(f"Kết quả: y = {w:.2f} * x + {b:.2f}")
```

```python
# Áp dụng vào bài toán
df = pd.read_csv('Data/Advertising.csv')

x = df['TV'].values
y = df['Sales'].values

b, w = min_least_squares(x, y)

print("--- Kết quả tính toán trên tập dữ liệu Advertising ---")
print(f"Hệ số góc (w - TV): {w:.5f}")
print(f"Hệ số chặn (b):     {b:.5f}")
print(f"Phương trình hồi quy tuyến tính: Sales = {w:.5f} * TV + {b:.5f}")
```

<h1 style= "font-size : 30px; ">3. Viết công thức Gradient Descent (GD) tính $w, b$ </h1>

Hàm mất mát (Mean Squared Error - MSE) cho hồi quy tuyến tính đơn biến được định nghĩa là: $$E(b, w) = \frac{1}{n} \sum_{i=1}^n \left( y_i - (b + w x_i) \right)^2$$

3.1. Tính đạo hàm riêng (Gradient)
Để tìm hướng giảm hàm mất mát, ta tính đạo hàm riêng của $E$ theo từng tham số $b$ và $w$:

$$\frac{\partial E}{\partial b} = -\frac{2}{n} \sum_{i=1}^n \left( y_i - (b + w x_i) \right)$$

$$\frac{\partial E}{\partial w} = -\frac{2}{n} \sum_{i=1}^n \left( y_i - (b + w x_i) \right) x_i$$

3.2. Công thức cập nhật tham số
Tại mỗi bước lặp, ta cập nhật $w$ và $b$ ngược hướng với đạo hàm bằng cách trừ đi một lượng tỷ lệ với tốc độ học $\eta > 0$:

$$b_{\text{mới}} = b_{\text{cũ}} - \eta \frac{\partial E}{\partial b} = b_{\text{cũ}} + \frac{2\eta}{n} \sum_{i=1}^n \left( y_i - (b_{\text{cũ}} + w_{\text{cũ}} x_i) \right)$$

$$w_{\text{mới}} = w_{\text{cũ}} - \eta \frac{\partial E}{\partial w} = w_{\text{cũ}} + \frac{2\eta}{n} \sum_{i=1}^n \left( y_i - (b_{\text{cũ}} + w_{\text{cũ}} x_i) \right) x_i$$

<h1 style= "font-size : 30px; ">4. Viết công thức Stochastic Gradient Descent (SGD) tính $w, b$ </h1>

Thay vì tính toán trên toàn bộ tập dữ liệu cùng một lúc (như Batch Gradient Descent), SGD chỉ chọn ngẫu nhiên một mẫu dữ liệu $(x_i, y_i)$ ở mỗi bước để cập nhật tham số.

4.1. Hàm mất mát trên một mẫu thứ $i$
$$E_i = (y_i - (b + w x_i))^2$$

4.2. Đạo hàm riêng trên một mẫu thứ $i$
$$\frac{\partial E_i}{\partial b} = -2(y_i - (b + w x_i))$$

$$\frac{\partial E_i}{\partial w} = -2(y_i - (b + w x_i)) x_i$$

4.3. Công thức cập nhật tham số SGD
Tại mỗi bước lặp, ta cập nhật $w$ và $b$ theo mẫu $(x_i, y_i)$ được chọn:

$$b_{\text{mới}} = b_{\text{cũ}} + 2\eta (y_i - (b_{\text{cũ}} + w_{\text{cũ}} x_i))$$

$$w_{\text{mới}} = w_{\text{cũ}} + 2\eta (y_i - (b_{\text{cũ}} + w_{\text{cũ}} x_i)) x_i$$

<h1 style= "font-size : 30px; ">5. Viết công thức tính $R^2$ </h1>

Hệ số xác định $R^2$ (R-squared) dùng để đo lường mức độ phù hợp của mô hình hồi quy đối với tập dữ liệu thực tế.

5.1. Công thức tổng quát
$$R^2 = 1 - \frac{RSS}{TSS}$$

Trong đó:

RSS (Residual Sum of Squares - Tổng bình phương phần dư): Đo lượng biến thiên không được giải thích bởi mô hình. $$RSS = \sum_{i=1}^n (y_i - \hat{y}_i)^2$$
TSS (Total Sum of Squares - Tổng bình phương toàn phần): Đo lượng biến thiên tổng thể của dữ liệu thực tế so với giá trị trung bình $\bar{y}$. $$TSS = \sum_{i=1}^n (y_i - \bar{y})^2$$
5.2. Ý nghĩa của giá trị $R^2$
$R^2 \approx 1$: Mô hình giải thích được phần lớn sự biến thiên của dữ liệu (mô hình rất tốt).
$R^2 \approx 0$: Mô hình hầu như không giải thích được sự biến thiên của dữ liệu (mô hình kém hiệu quả).
